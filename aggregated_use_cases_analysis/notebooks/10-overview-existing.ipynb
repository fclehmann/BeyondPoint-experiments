```python
with open("/data/horse/ws/s4610340-isml/dev/statistics-in-machine-learning/dashboard/services/isml_db_init/simpsons.json") as fp:
   data["simpsons"] = json.load(fp)

for e in data["simpsons"]:
   print(e["experiment"]) 
```

```python
from pathlib import Path
from typing import Any
import tomlkit

def read_ec(exp_path: Path) -> dict[str, Any]:
    return tomlkit.parse(next(exp_path.glob("*.toml")).read_text())

for e in data["simpsons"]:
    path_e = Path(e["experiments_path"])
    path_results = path_e / "results"
    ec = read_ec(path_e)

    print(path_e)

    if "_hopt_" in str(path_e):
        n_runs_must = int(ec["DATA"]["max_evals"])*len(ec["DATA"]["seed"])
    else:
        n_runs_must = min(int(ec["DATA"]["max_evals"]), len(ec["DIMENSIONS"][0]["options"]))

    n_runs = len(list(path_results.glob("./*/result-*.json")))
    
    print(n_runs / n_runs_must)
```
