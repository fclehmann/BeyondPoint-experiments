```python
import pandas as pd
import plotly.express as px
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_style("darkgrid")
```

```python
import json

with open("/data/horse/ws/s4610340-isml/dev/statistics-in-machine-learning/dashboard/services/isml_db_init/cifar10_clf.json", "r") as fp:
    data_cifar10 = json.load(fp)
 
```

```python
fig, axs = plt.subplots(nrows=len(data_cifar10), ncols=1, figsize=(8, 2*len(data_cifar10)), sharex=True)
for ax, experiment in zip(axs, data_cifar10):
    sns.histplot(x=experiment["metric_values"]["val_MulticlassAccuracy"], ax=ax)
    ax.set_title(experiment["legend"])
fig.tight_layout()
```
