```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt
```

```python
p_exp = Path("/data/horse/ws/s4610340-isml/cifar10_clf/experiments_v2/008_cifar10_clf_hopt_tpe_SGDwlrs_bs128_s")
```

```python
r = {}
for p_hopt_seed in p_exp.glob("./hopt/*"):
    if (p_hopt_seed / "trials.csv").exists():
        df = pd.read_csv(p_hopt_seed / "trials.csv")
        if len(df.dropna()) == 200:
            hopt_seed = df["user_attrs_hopt_seed"].iloc[0]
            r[hopt_seed] = df["value"].min()
    else:
        print(f"Warning: {p_hopt_seed.stem} was not successfull")

r
```
