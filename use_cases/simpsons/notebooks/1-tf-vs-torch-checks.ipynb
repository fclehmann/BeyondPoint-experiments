```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import pickle
sns.set_theme()
```

```python
with open("/home/egor/Projects/Pycharm/statistics-in-machine-learning/inferentialStatisticsMLmetrics/projects/simpsons_project/data/tf_labels/train_labels.txt", "rb") as f:
    tf_train_labels = pickle.load(f)
with open("/home/egor/Projects/Pycharm/statistics-in-machine-learning/inferentialStatisticsMLmetrics/projects/simpsons_project/data/tf_labels/val_labels.txt", "rb") as f:
    tf_test_labels = pickle.load(f)
with open("/home/egor/Projects/Pycharm/statistics-in-machine-learning/inferentialStatisticsMLmetrics/projects/simpsons_project/data/torch_labels/train_labels.txt", "rb") as f:
    torch_train_labels = pickle.load(f)
with open("/home/egor/Projects/Pycharm/statistics-in-machine-learning/inferentialStatisticsMLmetrics/projects/simpsons_project/data/torch_labels/val_labels.txt", "rb") as f:
    torch_test_labels = pickle.load(f)
```

```python
def create_df(xs, source, split):
    df = pd.DataFrame(xs, columns=["y"])
    df["source"] = source
    df["split"] = split
    return df

DF = pd.concat([
    create_df(tf_train_labels, "tf", "train"),
    create_df(tf_test_labels, "tf", "test"),
    create_df(torch_train_labels, "torch", "train"),
    create_df(torch_test_labels, "torch", "test"),
])

sns.displot(DF, x="y", hue="source", multiple="dodge", fill=True, col="split", height=8, facet_kws=dict(sharey=False))

```
